# List of Dictionaries

```python
#Store the expenses as a list of dictionaries


expenses = [ 
    {"category": "Groceries", "item": "Food", "amount": 120000},
    {"category": "Transport", "item": "Fuel", "amount": 80000},
    {"category": "Housing", "item": "Rent", "amount": 250000},
    {"category": "Shopping", "item": "clothes", "amount": 80000},
    {"category": "School", "item": "Courses", "amount":60000 },
    {"category": "Fitness", "item": "Gym", "amount": 15000},
    {"category": "Insurance", "item": "Health", "amount": 50000},
    {"category": "Investment", "item": "Dangote refinery", "amount": 100000},
    {"category": "Skincare", "item": "Make-up", "amount": 60000},
    {"category": "Travelling", "item": "Bali", "amount": 1500000}]


budgets = {"Groceries": 150000, "Transport": 50000, "Housing": 250000, "Shopping": 60000, "School": 60000, "Fitness": 15000, "Insurance": 25000, "Investment": 100000, "Skincare": 60000, "Travelling": 0 }


totals = {} # Total the spending per category. Use a loop and a dictionary to add up the amounts as you go.
for e in expenses:
    print(e["category"], e["amount"])
```

```python
#Flag categories that exceed their budget. Use if / else to compare each total with its budget.

totals = {}
for e in expenses:
    cat = e["category"]
    totals[cat] = totals.get(cat, 0) + e["amount"]

for cat in budgets:
    spent = totals.get(cat, 0)
    status = "OVER" if spent > budgets[cat] else "OK"
    total_amount = spent - budgets[cat]
    
#Use f-strings to show the category, amount spent, budget and status on one line each.
    print(f"{cat:>10} spent {spent:>10,} of {budgets[cat]:>10,} {status:>10} by {total_amount:>10}")

# Here I used the lecture notes for the calculation
```
