# Day 2 : In-Class Task

```python
''' Collecting of Price List'''

item1 = input("Item 1 name: ")
price1 = float(input("Price per unit (NGN): "))
qty1 = int(input("Quantity: "))

item2 = input("Item 2 name: ")
price2 = float(input("Price per unit (NGN): "))
qty2 = int(input("Quantity: "))

item3 = input("Item 3 name: ")
price3 = float(input("Price per unit (NGN): "))
qty3 = int(input("Quantity: "))
```

```python
'''Preparing Discounts and Receipt'''

total_price = price1 * qty1 + price2 * qty2
if total_price >= 10000:
 price_discount = total_price * 0.10
else:
 price_discount = 0
    
print("------ RECEIPT ------")
print(f"{item1} x{qty1}: NGN {price1 * qty1:,.2f}")
print(f"{item2} x{qty2}: NGN {price2 * qty2:,.2f}")
print(f"{item3} x{qty3}: NGN {price3 * qty3:,.2f}")

print(f"Discount: NGN {price_discount:,.2f}")
print(f"TOTAL: NGN {total_price - price_discount:,.2f}")
```
